```python
LAST_NAME="FUENTES";STUDENT_ID="TUPM-26-1908";FAVORITE_ARTIST="MICHAEL_JACKSON"
SEED_DIGIT=int(STUDENT_ID[-1]); ID_SUM=sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH=len(LAST_NAME); ARTIST_LENGTH=len(FAVORITE_ARTIST)
ARTIST_SUM=sum(ord(c.upper())-64 for c in FAVORITE_ARTIST if c.isalpha())

n=20
original = list(range(100, 100 + n*SEED_DIGIT, SEED_DIGIT))
rot = SEED_DIGIT % n
rotated = original[rot:] + original[:rot]
target = original[(ID_SUM + ARTIST_SUM) % n]

def lin(a, t):
    c = 0
    for i, v in enumerate(a):
        c += 1
        if v== t: return i,c
    return -1, c

def rot_bin(a,t):
    lo, hi, c = 0, len(a)-1,0
    while lo <= hi:
        mid = (lo+hi)//2; c += 1
        if a[mid] == t: return mid,  c
        if a[lo] <= a[mid]:
            if a[lo] <= t < a[mid]: hi = mid-1
            else: lo = mid+1
        else:
            if a[mid] < t <= a[hi]: lo = mid+1
            else: hi = mid-1
    return -1, c

i1, c1 = lin(rotated, target);i2, c2 = rot_bin(rotated, target)
print("Original:", original); print("Rotated:", rotated)
print(f"Target={target} | Linear: idx={i1}, c,p={c1} | RotatedBinary: idx={i2}, cmp={c2}")
print("Complexity -> Linear: O(n) time, O(1) space | RotatedBinary: O(log n) time, O(1) space")
```
